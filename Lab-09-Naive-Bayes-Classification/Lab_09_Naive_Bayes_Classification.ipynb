```python
# Step 1: Import required libraries
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

# Step 2: Load the Iris dataset
iris = load_iris()

X = iris.data
y = iris.target

# Step 3: Display dataset information
print("Iris Dataset")
print("Number of samples:", len(X))
print("Number of features:", X.shape[1])
print("Target classes:", iris.target_names)

# Step 4: Split training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0, stratify=y
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))

# Step 5: Create Gaussian Naive Bayes classifier
model = GaussianNB()

# Step 6: Train the classifier
model.fit(X_train, y_train)

# Step 7: Predict test data
y_pred = model.predict(X_test)

# Step 8: Display actual and predicted values
print("\nActual and Predicted Classes:")
for i in range(len(y_test)):
    actual = iris.target_names[y_test[i]]
    predicted = iris.target_names[y_pred[i]]
    print("Actual:", actual, "| Predicted:", predicted)

# Step 9: Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print("\nClassification Accuracy:", accuracy * 100, "%")

# Step 10: Display confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# Step 11: Display classification report
print("\nClassification Report:")
print(classification_report(
    y_test, y_pred, target_names=iris.target_names
))
```
