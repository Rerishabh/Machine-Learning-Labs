##**Code 1: Basic Polynomial **SVM****

```python
# Polynomial SVM Classification

from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Load dataset (Iris example)
data = datasets.load_iris()
X = data.data
y = data.target

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Feature scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Polynomial SVM
poly_svm = SVC(kernel='poly', degree=3, C=1.0, gamma='scale')
poly_svm.fit(X_train, y_train)

# Prediction
y_pred = poly_svm.predict(X_test)

# Performance
print("Polynomial SVM Accuracy:", accuracy_score(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

##Code 2: Modified SVM (Final Experiment)

```python
# SVM Hyperparameter Tuning using Linear and RBF Kernels

# Step 1: Import libraries
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Step 2: Load a different dataset
data = load_breast_cancer()

X = data.data
y = data.target

print("Breast Cancer Dataset")
print("Number of samples:", X.shape[0])
print("Number of features:", X.shape[1])
print("Target classes:", data.target_names)

# Step 3: Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# Step 4: Feature scaling
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Step 5: Define C and gamma values
C_values = [0.01, 1, 10, 100]
gamma_values = [0.01, 1, 10, 100]

kernels = ['linear', 'rbf']

best_accuracy = 0
best_model = None
best_parameters = None

# Step 6: Try different combinations
for kernel in kernels:
    for c in C_values:
        for g in gamma_values:

            model = SVC(kernel=kernel, C=c, gamma=g)

            model.fit(X_train, y_train)

            y_pred = model.predict(X_test)

            accuracy = accuracy_score(y_test, y_pred)

            print("Kernel:", kernel, "C:", c,
                  "Gamma:", g, "Accuracy:", accuracy * 100)

            if accuracy > best_accuracy:
                best_accuracy = accuracy
                best_model = model
                best_parameters = (kernel, c, g)

# Step 7: Display best results
print("\nBest Parameters:")
print("Kernel:", best_parameters[0])
print("C:", best_parameters[1])
print("Gamma:", best_parameters[2])

print("\nBest Accuracy:", best_accuracy * 100, "%")

# Step 8: Evaluate the best model
y_best_pred = best_model.predict(X_test)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_best_pred))

print("\nClassification Report:")
print(classification_report(
    y_test, y_best_pred, target_names=data.target_names
))
```
