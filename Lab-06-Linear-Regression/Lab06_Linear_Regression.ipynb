# Question 6: Linear Regression Model Implementation and Evaluation using MAE, MSE, RMSE, and R² Score


```python
# Import necessary libraries
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Given sample data
X = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)
Y = np.array([8, 10, 12, 14, 16, 18])

# Create and train the Linear Regression model
model = LinearRegression()
model.fit(X, Y)

# Predict values using Linear Regression
Y_pred = model.predict(X)

# Display slope and intercept
print("Linear Regression Results")
print("-------------------------")
print("Slope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)

# Display actual and predicted values
print("\nActual and Predicted Values:")
print("Sample\tActual Y\tPredicted Y")

for i in range(len(Y)):
    print(f"{X[i][0]}\t{Y[i]}\t\t{Y_pred[i]:.2f}")

# (a) Calculate Mean Absolute Error
MAE = mean_absolute_error(Y, Y_pred)
print("\n(a) Mean Absolute Error (MAE):", round(MAE, 2))

# (b) Calculate MSE and RMSE
MSE = mean_squared_error(Y, Y_pred)
RMSE = np.sqrt(MSE)

print("(b) Mean Squared Error (MSE):", round(MSE, 2))
print("    Root Mean Squared Error (RMSE):", round(RMSE, 2))

# (c) Calculate R2 Score
R2 = r2_score(Y, Y_pred)
print("(c) R2 Score:", round(R2, 2))

# (d) Comment on model performance
print("\n(d) Model Performance:")
if R2 == 1.0:
    print("The model fits the given data perfectly.")
    print("It explains 100% of the variation in the target values.")
else:
    print("The model performance is evaluated using the R2 score.")
```
